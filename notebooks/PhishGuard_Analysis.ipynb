### Investigate duplicates before cleaning

```python

import pandas as pd

df111 = pd.read_csv("dataset_full.csv")

feature_cols = [c for c in df111.columns if c != "phishing"]

# Find feature combinations associated with multiple labels
label_counts = (
    df111.groupby(feature_cols, dropna=False)["phishing"]
    .agg(["count", "nunique"])
    .reset_index()
)

conflicts = label_counts[label_counts["nunique"] > 1]

print("Conflicting feature combinations:", len(conflicts))
print("\nConflicting combinations and their feature values:")
display(conflicts)

# Show the original records for those combinations
if len(conflicts) > 0:
    conflicting_rows = df111.merge(
        conflicts[feature_cols],
        on=feature_cols,
        how="inner"
    )
    print("\nOriginal rows with conflicting labels:")
    display(conflicting_rows.sort_values(feature_cols).head(30))

# Inspect repeated records and label agreement
print("\nExact duplicate rows beyond first occurrence:",
      df111.duplicated().sum())

print("\nClass counts before any cleaning:")
print(df111["phishing"].value_counts())
```

What we learned
- 88,647 records and 111 features.
- No missing values and no duplicate column names.
- 65.43% legitimate, 34.57% phishing — moderately imbalanced, but we can start with stratified splitting and suitable metrics.
- 1,438 exact duplicate rows beyond their first occurrences.
- 6 identical feature combinations with conflicting labels.

The conflicting records have feature values such as ttl_hostname = -1, qty_redirects = -1, and qty_ip_resolved = -1. These may represent unavailable or unknown information, depending on the dataset's encoding. We shouldn't treat -1 as missing automatically without confirming that convention.

### Cleaning

```python

import pandas as pd

df111 = pd.read_csv("dataset_full.csv")

target = "phishing"
features = [c for c in df111.columns if c != target]

# Step 1: Identify feature combinations with conflicting labels
label_counts = (
    df111.groupby(features, dropna=False)[target]
    .nunique()
)

conflicting_features = label_counts[label_counts > 1].index

print("Conflicting feature combinations:", len(conflicting_features))

# Step 2: Flag all records belonging to conflicting combinations
feature_index = pd.MultiIndex.from_frame(df111[features])
conflict_index = pd.MultiIndex.from_tuples(
    list(conflicting_features),
    names=features
)
conflict_mask = feature_index.isin(conflict_index)

print("Records excluded due to conflicting labels:",
      int(conflict_mask.sum()))

# Step 3: Exclude those ambiguous combinations
df_clean = df111.loc[~conflict_mask].copy()

# Step 4: Remove exact duplicate rows
before_dedup = len(df_clean)
df_clean = df_clean.drop_duplicates().reset_index(drop=True)

print("\nRows before cleaning:", len(df111))
print("Rows after conflict removal and deduplication:", len(df_clean))
print("Rows removed:", len(df111) - len(df_clean))

print("\nCleaned class distribution:")
print(df_clean[target].value_counts())

print("\nCleaned class percentages:")
print((df_clean[target].value_counts(normalize=True) * 100).round(2))

print("\nMissing values:", df_clean.isnull().sum().sum())
print("Remaining exact duplicates:", df_clean.duplicated().sum())
```

Final cleaning results

The cleaned class distribution is:
- Legitimate (0): 56,706 — 65.03%
- Phishing (1): 30,491 — 34.97%
The dataset remains moderately imbalanced. That's okay—we'll preserve the class proportions during splitting and assess more than accuracy.

## Step 3 — Train-Test Split
(for the 111-feature dataset)

```python

from sklearn.model_selection import train_test_split

def load_and_prepare_data_111(df_clean):

    # Separate features and target
    X = df_clean.drop(columns=["phishing"])
    y = df_clean["phishing"]

    # Train-test split: same 70:30 ratio
    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y,
        train_size=0.70,
        test_size=0.30,
        stratify=y,
        random_state=42
    )

    return X_train, X_test, y_train, y_test


X_train, X_test, y_train, y_test = load_and_prepare_data_111(df_clean)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)

print("\nTraining class distribution:")
print(y_train.value_counts())

print("\nTesting class distribution:")
print(y_test.value_counts())
```

The class distribution is also preserved closely across both sets.
- Training: 39,694 legitimate (class 0), 21,343 phishing (class 1).
- Testing: 17,012 legitimate (class 0), 9,148 phishing (class 1).

This confirms that stratify=y has maintained approximately the same class proportions in both sets. Your random_state=42 also makes the split reproducible.

##Step 4 - Train baseline ML models

(111 Features dataset)



```python

# STEP 4: Train Baseline ML Models (111 Features)

import pandas as pd
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)
from xgboost import XGBClassifier

models = {
    "Logistic Regression": Pipeline([
        ("scaler", StandardScaler()),
        ("model", LogisticRegression(max_iter=2000, random_state=42))
    ]),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "Random Forest": RandomForestClassifier(
        n_estimators=100, random_state=42, n_jobs=-1
    ),
    "XGBoost": XGBClassifier(
        n_estimators=100,
        random_state=42,
        eval_metric="logloss",
        n_jobs=-1
    )
}

results = []

for name, model in models.items():
    print(f"Training {name}...")
    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)[:, 1]

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred, zero_division=0),
        "Recall": recall_score(y_test, y_pred, zero_division=0),
        "F1-Score": f1_score(y_test, y_pred, zero_division=0),
        "ROC-AUC": roc_auc_score(y_test, y_prob)
    })

results_df = pd.DataFrame(results)
print("\nBaseline Model Results:")
display(results_df.round(4))
```

On the 48-feature dataset, XGBoost leads on accuracy and F1-score, while Random Forest leads on those two metrics on the 111-feature dataset.

##Step 5 -  SHAP-based feature reduction
### Step 5A — SHAP analysis for the 48-feature dataset

```python

# STEP 5A: SHAP-Based Feature Selection (48 Features)

import pandas as pd
import numpy as np
import shap
from xgboost import XGBClassifier

# 1. Train XGBoost using training data only
xgb_48 = XGBClassifier(
    n_estimators=100,
    random_state=42,
    eval_metric="logloss",
    n_jobs=-1
)

xgb_48.fit(X_train, y_train)

# 2. Calculate SHAP values on training data
explainer_48 = shap.TreeExplainer(xgb_48)
shap_values_48 = explainer_48.shap_values(X_train)

# Handle binary-classification SHAP output
if isinstance(shap_values_48, list):
    shap_matrix_48 = shap_values_48[1]
elif isinstance(shap_values_48, np.ndarray) and shap_values_48.ndim == 3:
    shap_matrix_48 = shap_values_48[:, :, 1]
else:
    shap_matrix_48 = shap_values_48

# 3. Calculate mean absolute SHAP importance
importance_48 = pd.DataFrame({
    "Feature": X_train.columns,
    "Mean_Abs_SHAP": np.abs(shap_matrix_48).mean(axis=0)
}).sort_values("Mean_Abs_SHAP", ascending=False)

# 4. Select the top 25 features
top_25_features = importance_48.head(25)["Feature"].tolist()

print("Top 25 features selected:")
display(importance_48.head(25).reset_index(drop=True))

print("\nNumber of selected features:", len(top_25_features))
```

Step 5A — Conclusion
- XGBoost was trained using the training partition, and SHAP values were used to rank the 48 features.
- The top 25 features were selected based on their mean absolute SHAP values.
- qty_slash_url was the most influential feature, with a mean absolute SHAP value of 2.241943, followed by time_domain_activation and directory_length.
- The reduced dataset will contain 25 features instead of 48, a reduction of approximately 47.9%.

```python

# STEP 6A - Setup: Load and split the original 48-feature dataset

import pandas as pd
from sklearn.model_selection import train_test_split

df48 = pd.read_csv("/content/Phishing_Legitimate_full.csv")

X48 = df48.drop(columns=["id", "CLASS_LABEL"])
y48 = df48["CLASS_LABEL"]

X_train_48, X_test_48, y_train_48, y_test_48 = train_test_split(
    X48,
    y48,
    train_size=0.70,
    test_size=0.30,
    stratify=y48,
    random_state=42
)

print("Training features:", X_train_48.shape)
print("Testing features:", X_test_48.shape)
```

```python

# Calculate SHAP importance for the original 48-feature dataset

import numpy as np
import shap
from xgboost import XGBClassifier

xgb_48 = XGBClassifier(
    n_estimators=100,
    random_state=42,
    eval_metric="logloss",
    n_jobs=-1
)

xgb_48.fit(X_train_48, y_train_48)

explainer_48 = shap.TreeExplainer(xgb_48)
shap_values_48 = explainer_48.shap_values(X_train_48)

if isinstance(shap_values_48, list):
    shap_matrix_48 = shap_values_48[1]
elif isinstance(shap_values_48, np.ndarray) and shap_values_48.ndim == 3:
    shap_matrix_48 = shap_values_48[:, :, 1]
else:
    shap_matrix_48 = shap_values_48

importance_48 = pd.DataFrame({
    "Feature": X_train_48.columns,
    "Mean_Abs_SHAP": np.abs(shap_matrix_48).mean(axis=0)
}).sort_values("Mean_Abs_SHAP", ascending=False)

top_25_features_48 = importance_48.head(25)["Feature"].tolist()

display(importance_48.head(25).reset_index(drop=True))
print("Selected features:", len(top_25_features_48))
```

```python

# STEP 6A: Evaluate All Four Models on the Reduced 48-Feature Dataset

import pandas as pd
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score
)
from xgboost import XGBClassifier

# 1. Keep only the selected top 25 features
X_train_25 = X_train_48[top_25_features_48]
X_test_25 = X_test_48[top_25_features_48]

print("Reduced training shape:", X_train_25.shape)
print("Reduced testing shape:", X_test_25.shape)

# 2. Define the four models
models_25 = {
    "Logistic Regression": Pipeline([
        ("scaler", StandardScaler()),
        ("model", LogisticRegression(max_iter=2000, random_state=42))
    ]),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "Random Forest": RandomForestClassifier(
        n_estimators=100, random_state=42, n_jobs=-1
    ),
    "XGBoost": XGBClassifier(
        n_estimators=100,
        random_state=42,
        eval_metric="logloss",
        n_jobs=-1
    )
}

# 3. Train and evaluate each model
results_25 = []

for name, model in models_25.items():
    print(f"Training {name}...")
    model.fit(X_train_25, y_train_48)

    y_pred = model.predict(X_test_25)
    y_prob = model.predict_proba(X_test_25)[:, 1]

    results_25.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test_48, y_pred),
        "Precision": precision_score(y_test_48, y_pred, zero_division=0),
        "Recall": recall_score(y_test_48, y_pred, zero_division=0),
        "F1-Score": f1_score(y_test_48, y_pred, zero_division=0),
        "ROC-AUC": roc_auc_score(y_test_48, y_prob)
    })

results_25_df = pd.DataFrame(results_25)

print("\nResults: Reduced 48-Feature Dataset (Top 25 Features)")
display(results_25_df.round(4))
```

## Step 8 - Computational Cost Analysis

```python

# Find existing dataset and model variables in the notebook

keywords = [
    "X_train", "X_test", "y_train", "y_test",
    "X_train_reduced", "X_test_reduced",
    "top_25", "top_50", "model", "clf"
]

for name, value in sorted(globals().items()):
    if any(keyword.lower() in name.lower() for keyword in keywords):
        if not name.startswith("_"):
            try:
                shape = value.shape
                print(f"{name}: shape={shape}")
            except AttributeError:
                if hasattr(value, "predict") or isinstance(value, dict):
                    print(f"{name}: type={type(value).__name__}")
```

We'll compare:

1. Training time — time required to train the model.
2. Prediction time — time to classify the test set.
3. Single-URL latency — time taken to predict one sample.
4. Model size — storage required for the trained model.
5. Predictive performance — accuracy and F1-score, to ensure efficiency doesn't come at the cost of detection quality.

```python

import os
import time
import tempfile
import statistics
import joblib
import numpy as np
import pandas as pd

from sklearn.base import clone
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, f1_score
from xgboost import XGBClassifier

# 1. Existing train/test splits
datasets = {
    "48 Original": (X_train_48, X_test_48, y_train_48, y_test_48),
    "48 Reduced (25)": (X_train_25, X_test_25, y_train_48, y_test_48),
    "111 Original": (X_train, X_test, y_train, y_test),
    "111 Reduced (50)": (X_train_50, X_test_50, y_train, y_test),
}

# 2. Consistent benchmark model configurations
models = {
    "Logistic Regression": LogisticRegression(max_iter=1000),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "Random Forest": RandomForestClassifier(
        n_estimators=100, random_state=42, n_jobs=-1
    ),
    "XGBoost": XGBClassifier(
        random_state=42, eval_metric="logloss", n_jobs=-1
    ),
}

results = []

for dataset_name, (Xtr, Xte, ytr, yte) in datasets.items():
    ytr = np.asarray(ytr).ravel()
    yte = np.asarray(yte).ravel()

    for model_name, template in models.items():
        print(f"Running: {dataset_name} | {model_name}")

        # Training time: median of 3 independent fits
        train_times = []
        fitted = None

        for _ in range(3):
            clf = clone(template)
            start = time.perf_counter()
            clf.fit(Xtr, ytr)
            train_times.append(time.perf_counter() - start)
            fitted = clf

        train_time = statistics.median(train_times)

        # Warm-up before prediction timing
        first_row = Xte.iloc[[0]] if hasattr(Xte, "iloc") else Xte[:1]
        fitted.predict(first_row)

        # Full test-set prediction time: median of 5 runs
        batch_times = []
        for _ in range(5):
            start = time.perf_counter()
            predictions = fitted.predict(Xte)
            batch_times.append(time.perf_counter() - start)

        batch_time = statistics.median(batch_times)

        # Single-row prediction latency: median over up to 100 rows
        single_times = []
        n = min(100, len(Xte))

        for i in range(n):
            row = Xte.iloc[[i]] if hasattr(Xte, "iloc") else Xte[i:i+1]
            start = time.perf_counter()
            fitted.predict(row)
            single_times.append((time.perf_counter() - start) * 1000)

        single_latency_ms = statistics.median(single_times)

        # Serialized model size
        with tempfile.TemporaryDirectory() as temp_dir:
            path = os.path.join(temp_dir, "model.joblib")
            joblib.dump(fitted, path)
            model_size_mb = os.path.getsize(path) / (1024 ** 2)

        results.append({
            "Dataset": dataset_name,
            "Model": model_name,
            "Feature_Count": Xtr.shape[1],
            "Training_Time_s": train_time,
            "Full_Test_Prediction_s": batch_time,
            "Single_Row_Latency_ms": single_latency_ms,
            "Model_Size_MB": model_size_mb,
            "Accuracy_pct": accuracy_score(yte, predictions) * 100,
            "F1_pct": f1_score(yte, predictions, zero_division=0) * 100,
        })

benchmark_df = pd.DataFrame(results)

display(
    benchmark_df.sort_values(["Dataset", "Model"]).round(4)
)

benchmark_df.to_csv("phishguard_step8_benchmark.csv", index=False)
print("Saved results to phishguard_step8_benchmark.csv")
```

What do these results tell us?
- 48-feature dataset: XGBoost with 25 features retains strong performance. Accuracy drops only from 98.77% to 98.60%, while training time increases slightly in this run.
- 111-feature dataset: XGBoost with 50 features achieves 97.03% accuracy, marginally higher than the original 96.93%. Its measured training time falls from 2.229 seconds to 1.245 seconds.
- Random Forest: It performs well, but its model files are substantially larger and its single-row prediction latency is higher than XGBoost's in these measurements.
- Decision Tree: It is fast and lightweight, but its accuracy is lower than the ensemble models.
- Logistic Regression: Its training times are unusually high on the 111-feature dataset, and the convergence warnings mean those timings and metrics should be treated cautiously.

One important distinction: these timings depend on the Colab runtime, its load, and the model settings. They are useful for comparing these configurations in this run, not universal performance guarantees.
